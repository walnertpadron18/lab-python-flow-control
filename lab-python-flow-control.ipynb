```python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
```

```python
inventory = {}
```

```python
for i in products:
    inventory[i] = int(input(f"Cuantas unidades necesita de {i}? "))
```

```python
customer_orders = set()
```

```python
continuar = "yes"
while continuar == "yes":
    continuar = input("¿Quieres comprar un articulo? (yes/no): ").lower()
    if continuar == "yes":
        product = input("Ingrese el articulo: ")
        if product in inventory:
            inventory[product] -= 1
            customer_orders.add(product)
        else:
            print("Ese artículo no está en el inventario.")
    else:
        print("Gracias por su compra")
```
